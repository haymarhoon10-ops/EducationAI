```python
# 2 sayeds(wlad alrsool) in project = A+

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from imblearn.over_sampling import SMOTE   # if colab says it is missing, run: !pip install imbalanced-learn

import warnings
warnings.filterwarnings("ignore")

RANDOM_STATE = 42
sns.set_style("whitegrid")
pd.set_option("display.max_columns", 50)

DATA_PATH = "Project1-EducationDataset.csv"
df_raw = pd.read_csv(DATA_PATH)
df = df_raw.copy()   # keep the raw one untouched so we can compare before vs after later

print(df.shape)
display(df.head())

# MILESTONE 1: DATA CLEANING AND OUTLIER HANDLING

# ======================================================================
# 1.1 FIRST LOOK AT THE DATA
# ======================================================================
#
# before touching anything we need to see what we're dealing with: types, ranges, and what looks
# weird.

df.info()

display(df.describe().T)

# ok so a few things already jump out from describe():
# - numeric_grade goes from -10 to 110, a grade can't be outside 0 to 100
# - prior_gpa has -1 and 5.0, gpa is on a 0 to 4 scale so those are impossible
# - age max is 65 in a dataset where most students are around 18 to 25
# - study_hours goes up to almost 120 which is way too much
#
# we'll deal with each of these below.

# quick look at the categorical columns to catch weird labels
cat_cols = ["semester", "subject_area", "grade_letter", "gender", "year_of_study"]
for col in cat_cols:
    print(f"--- {col} ---")
    print(df[col].value_counts(), "\n")

print("unique course codes:", df["course_code"].nunique())
print("unique course names:", df["course_name"].nunique())
print("unique instructors :", df["instructor"].nunique())

# grade_letter has an ERR label (780 rows) which obviously isn't a real grade. also
# extracurricular goes from 0 to 10 so it's actually a count, not a yes/no like the brief says.
# we'll treat it as a number.


# ======================================================================
# 1.2 MISSING VALUES
# ======================================================================
#
# checking how much is missing per column and in what %.

missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2)
})
display(missing[missing["missing_count"] > 0])

missing_cols = ["study_hours", "attendance_rate", "homework_avg", "final_exam"]
(df[missing_cols].isna().mean() * 100).plot(kind="bar", color="salmon", figsize=(6, 3))
plt.ylabel("% missing")
plt.title("missing values per column")
plt.show()

# so 4 columns have around 7% missing each: study_hours, attendance_rate, homework_avg and
# final_exam.
#
# 7% isn't huge but dropping every row with any missing value would throw away a big chunk of the
# data, so we're not doing that. plan is:
# - final_exam is our regression target, so we're not filling it in. if we made up the target the
#   model would just learn our guesses. we leave it as NaN for now and drop those rows only when
#   we train the regression model. we keep them for now because the other columns in those rows
#   are still useful (and we need them for the semester history in milestone 2)
# - study_hours, attendance_rate, homework_avg we fill in, but only after we fix the bad values
#   below, otherwise the medians would be pulled by the garbage numbers


# ======================================================================
# 1.3 INCONSISTENT AND INVALID RECORDS
# ======================================================================

# ======================================================================
# ERR GRADES AND NUMERIC_GRADE OUTSIDE 0 TO 100
# ======================================================================

err_rows = df["grade_letter"] == "ERR"
bad_numeric = (df["numeric_grade"] < 0) | (df["numeric_grade"] > 100)

print("ERR rows:", err_rows.sum())
print("numeric_grade outside 0-100:", bad_numeric.sum())
display(pd.crosstab(err_rows, bad_numeric, rownames=["is ERR"], colnames=["bad numeric_grade"]))

# the crosstab shows it's the exact same 780 rows. every ERR grade also has a broken
# numeric_grade, so we can't recover the real grade from either column. that's only about 0.5% of
# the data, so we just drop them. keeping them would give the classifier a fake 11th class and
# filling them in would basically be guessing the target.

df = df[df["grade_letter"] != "ERR"].copy()
print("rows left:", len(df))

# sanity check: does each letter grade match a clean numeric range now?
display(df.groupby("grade_letter")["numeric_grade"].agg(["min", "max", "count"]).sort_values("min", ascending=False))

# nice, every letter maps to its own clean band now (A = 93 to 100, A- = 90 to 93 ...). this also
# shows numeric_grade basically is the letter grade in number form, which is why we have to keep
# it out of the classification features later (that's leakage).


# ======================================================================
# PRIOR_GPA OUTSIDE 0 TO 4
# ======================================================================

bad_gpa = (df["prior_gpa"] < 0) | (df["prior_gpa"] > 4)
print("bad prior_gpa rows:", bad_gpa.sum())
print(df.loc[bad_gpa, "prior_gpa"].value_counts())

# only two values show up, exactly -1 and 5.0, which looks like placeholder/typo values, not real
# gpas. we don't wanna drop the whole row for this because the rest of the row is fine, so we
# turn them into NaN and fill them later with the student's own median gpa.

df.loc[bad_gpa, "prior_gpa"] = np.nan


# ======================================================================
# REPEATED STUDENT + SEMESTER + COURSE ROWS
# ======================================================================
#
# a student can't really take the same course twice in the same semester, so let's check for
# that.

key = ["student_id", "semester", "course_code"]
dupes = df[df.duplicated(key, keep=False)].sort_values(key)
print("rows involved:", len(dupes))
print("fully identical rows:", df.duplicated().sum())
display(dupes.head(6))

# so these aren't exact copies, the scores are a bit different between them. our take is that
# it's a data entry repeat, because a retake would be in a different semester, not the same one.
# keeping both would count the same course twice for that student and mess up the per semester
# averages we build in milestone 2.
#
# so we keep one row per student + semester + course, and we keep the one with the fewest missing
# values since that's the most complete record.

df["n_missing"] = df.isna().sum(axis=1)
df = (df.sort_values(key + ["n_missing"])
        .drop_duplicates(key, keep="first")
        .drop(columns="n_missing")
        .reset_index(drop=True))
print("rows after removing repeats:", len(df))
print("any repeats left?", df.duplicated(key).any())


# ======================================================================
# STATIC INFO SHOULD BE THE SAME FOR EACH STUDENT
# ======================================================================
#
# gender, scholarship and year_of_study shouldn't change between rows of the same student in this
# data, so we check that.

per_student = df.groupby("student_id")[["gender", "scholarship", "year_of_study", "age"]].nunique()
display((per_student > 1).sum())

# gender, scholarship and year_of_study are all consistent. but age has more than one value for
# some students, so let's look at one.

weird_id = per_student[per_student["age"] > 1].index[0]
display(df[df["student_id"] == weird_id][["student_id", "semester", "age"]].drop_duplicates().sort_values("semester"))

# yeah so this student is 18 in every semester except one row where it suddenly says 65. that's
# clearly a typo. we'll fix age in the outlier section below by replacing those jumps with the
# student's normal age.


# ======================================================================
# 1.4 OUTLIER DETECTION
# ======================================================================
#
# we use boxplots + the IQR rule (anything below Q1 - 1.5·IQR or above Q3 + 1.5·IQR is flagged).
# we went with IQR over z-score because z-score assumes the data is roughly normal and uses the
# mean and std, which the outliers themselves pull around. IQR uses quartiles so it doesn't get
# fooled by the extreme values it's trying to find.

num_cols = ["numeric_grade", "prior_gpa", "study_hours", "attendance_rate",
            "age", "homework_avg", "final_exam", "extracurricular", "credits"]

def iqr_outliers(data, col):
    # returns the lower fence, upper fence and how many values fall outside them
    q1, q3 = data[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((data[col] < low) | (data[col] > high)).sum()
    return low, high, n_out

rows = []
for col in num_cols:
    low, high, n_out = iqr_outliers(df, col)
    rows.append([col, round(low, 2), round(high, 2), n_out, round(n_out / len(df) * 100, 2)])

outlier_table = pd.DataFrame(rows, columns=["column", "lower_fence", "upper_fence", "n_outliers", "pct"])
display(outlier_table)

def plot_boxes(data, cols, title):
    fig, axes = plt.subplots(1, len(cols), figsize=(3 * len(cols), 4))
    for ax, col in zip(axes, cols):
        sns.boxplot(y=data[col], ax=ax, color="skyblue")
        ax.set_title(col)
        ax.set_ylabel("")
    fig.suptitle(title)
    plt.tight_layout()
    plt.show()

check_cols = ["study_hours", "age", "attendance_rate", "homework_avg", "final_exam", "extracurricular"]
plot_boxes(df, check_cols, "before outlier handling")

# looking closer at study_hours since the boxplot has that weird cluster at the top
plt.figure(figsize=(8, 3))
plt.hist(df["study_hours"].dropna(), bins=120, color="teal")
plt.xlabel("study_hours")
plt.title("study_hours distribution")
plt.show()

print(df["study_hours"].quantile([0.5, 0.9, 0.99, 0.999]))

# the IQR flags stuff in pretty much every column but we shouldn't just delete everything it
# flags. we looked at each one and decided case by case:
#
# - study_hours: the histogram shows the real data ends around 55 and then there's a separate
#   lump from about 75 to 120 with a gap in between. 80+ hours a week on one course isn't
#   realistic, so that lump is bad data, not hard working students. we set anything above 60 to
#   NaN and fill it like the other missing ones. what's left is real, but the long tail gets
#   capped at the 99th percentile (around 34 hours) so a few heavy studiers don't drag the
#   scaling around. capping keeps those students in the data, it just pulls their value in a bit.
# - age: the 55 / 60 / 65 values only show up as one off jumps for students who are 18 to 25
#   everywhere else (we saw that above). we replace them with that student's most common age.
# - final_exam, homework_avg, attendance_rate: the flagged values are just low scores, like
#   someone who got 40 on the final. that's real and honestly exactly the at-risk students we
#   want the model to catch, so we keep them.
# - extracurricular: 7 to 10 activities is rare but possible, it's a count. we keep it.
# - numeric_grade / prior_gpa: already fixed above, what's left is all inside the valid range.

# study_hours: impossible values become missing, then cap the real tail
df.loc[df["study_hours"] > 60, "study_hours"] = np.nan
cap = df["study_hours"].quantile(0.99)
df["study_hours"] = df["study_hours"].clip(upper=cap)
print("study_hours capped at:", round(cap, 2))

# age: swap the random jumps for the student's usual age
usual_age = df.groupby("student_id")["age"].agg(lambda x: x.mode().iloc[0])
df["age"] = np.where(df["age"] > 40, df["student_id"].map(usual_age), df["age"])
print("max age now:", df["age"].max())
print("students with more than one age now:", (df.groupby("student_id")["age"].nunique() > 1).sum())


# ======================================================================
# 1.5 FILLING IN THE MISSING VALUES
# ======================================================================
#
# now that the bad values are gone we can fill the gaps. we fill each one with the same student's
# median first, because a student's study habits and homework level are pretty consistent across
# their courses, so that's a way better guess than the whole dataset's median. if a student has
# no other values at all we fall back to the overall median.
#
# again, final_exam is not filled since it's the target.

def fill_with_student_median(data, col):
    student_med = data.groupby("student_id")[col].transform("median")
    data[col] = data[col].fillna(student_med)
    data[col] = data[col].fillna(data[col].median())   # fallback for students with nothing
    return data

for col in ["study_hours", "attendance_rate", "homework_avg", "prior_gpa"]:
    df = fill_with_student_median(df, col)

display(df.isna().sum()[df.isna().sum() > 0])

# only final_exam still has NaNs, which is what we want.


# ======================================================================
# 1.6 BEFORE VS AFTER
# ======================================================================

compare = pd.DataFrame({
    "before": [len(df_raw), df_raw.isna().sum().sum(), (df_raw["grade_letter"] == "ERR").sum(),
               df_raw["age"].max(), df_raw["study_hours"].max(),
               df_raw["prior_gpa"].min(), df_raw["prior_gpa"].max(),
               df_raw.duplicated(key).sum()],
    "after": [len(df), df.isna().sum().sum(), (df["grade_letter"] == "ERR").sum(),
              df["age"].max(), round(df["study_hours"].max(), 2),
              df["prior_gpa"].min(), df["prior_gpa"].max(),
              df.duplicated(key).sum()]
}, index=["rows", "total missing cells", "ERR grades", "max age", "max study_hours",
          "min prior_gpa", "max prior_gpa", "repeated student/semester/course"])
display(compare)

plot_boxes(df, check_cols, "after outlier handling")

# milestone 1 summary: we went from 154,314 to about 151.6k rows, so we lost less than 2% of the
# data. what we dropped was stuff we really couldn't trust (ERR rows and repeated entries).
# everything else got fixed in place instead of thrown away. the data's now in valid ranges, the
# missing values are filled (except the target) and the outliers we kept are real students, not
# errors. that's the point of this step: the models later learn from real patterns and not from
# typos.


# ======================================================================
# MILESTONE 2: PREPROCESSING AND FEATURE ENGINEERING
# ======================================================================

# ======================================================================
# 2.1 PUTTING THE SEMESTERS IN THE RIGHT ORDER
# ======================================================================
#
# semester is text like 2021Spring. if you sort that alphabetically you get 2020Fall before
# 2020Spring, which is backwards. so we turn it into a number: spring = first half of the year,
# fall = second half.

def semester_to_number(sem):
    year = int(sem[:4])
    term = 0 if "Spring" in sem else 1
    return (year - 2020) * 2 + term   # 2020Spring = 0, 2020Fall = 1, ... 2023Fall = 7

df["sem_idx"] = df["semester"].apply(semester_to_number)
display(df[["semester", "sem_idx"]].drop_duplicates().sort_values("sem_idx"))

df = df.sort_values(["student_id", "sem_idx", "course_code"]).reset_index(drop=True)
display(df[["student_id", "semester", "sem_idx", "course_code", "final_exam"]].head(10))


# ======================================================================
# 2.2 TIME SERIES FEATURES (LAGS AND ROLLING STATS)
# ======================================================================
#
# important thing here: each student takes several courses per semester (around 4). so "previous
# semester's final_exam" doesn't mean one row, it means the whole previous semester. that's why
# we first squash the data down to one row per student per semester (averages), build the lag and
# rolling features at that level, and then attach them back to every course row.

sem_level = (df.groupby(["student_id", "sem_idx"])
               .agg(sem_final_exam=("final_exam", "mean"),
                    sem_homework=("homework_avg", "mean"),
                    sem_study_hours=("study_hours", "mean"),
                    sem_attendance=("attendance_rate", "mean"),
                    n_courses=("course_code", "count"),
                    credit_load=("credits", "sum"))
               .reset_index())
display(sem_level.head())


# ======================================================================
# MAKING EACH STUDENT'S SEMESTERS CONTINUOUS
# ======================================================================
#
# some students skip a semester (they're in 2021Spring and 2022Spring but not 2021Fall, for
# example). if we just did shift(1), the "previous semester" for 2022Spring would be 2021Spring,
# which is wrong, it's a year back. so for each student we fill in every semester between their
# first and last one, and the missing semesters get NaN. that way the lag is always really the
# semester right before.

# first and last semester for every student
bounds = sem_level.groupby("student_id")["sem_idx"].agg(["min", "max"])

# every semester from first to last, then merge the real data onto it (gaps end up as NaN)
full_rows = [(student, sem) for student, row in bounds.iterrows()
             for sem in range(row["min"], row["max"] + 1)]
full_grid = pd.DataFrame(full_rows, columns=["student_id", "sem_idx"])

before = len(sem_level)
sem_level = full_grid.merge(sem_level, on=["student_id", "sem_idx"], how="left")
print("student-semester rows before:", before)
print("after filling gaps          :", len(sem_level))
print("gap semesters added         :", len(sem_level) - before)

# example of a student who had a gap so you can see the NaN row
gap_student = sem_level[sem_level["n_courses"].isna()]["student_id"].iloc[0]
display(sem_level[sem_level["student_id"] == gap_student])

# now the lag and rolling features. the key thing is we shift first, so the feature for a
# semester only uses the semesters before it and never the current one. if the rolling mean
# included the current semester's final_exam, we'd be giving the model the answer (leakage).
#
# - final_exam_prev_semester: average final exam last semester
# - homework_prev_semester: average homework last semester
# - final_exam_roll2 / final_exam_roll3: average final exam over the last 2 / 3 semesters
# - final_exam_trend: difference between the last two semesters, so is the student going up or
#   down

g = sem_level.groupby("student_id")

sem_level["final_exam_prev_semester"] = g["sem_final_exam"].shift(1)
sem_level["homework_prev_semester"] = g["sem_homework"].shift(1)

prev = g["sem_final_exam"].shift(1)
sem_level["final_exam_roll2"] = prev.groupby(sem_level["student_id"]).transform(lambda x: x.rolling(2, min_periods=1).mean())
sem_level["final_exam_roll3"] = prev.groupby(sem_level["student_id"]).transform(lambda x: x.rolling(3, min_periods=1).mean())

sem_level["final_exam_trend"] = g["sem_final_exam"].shift(1) - g["sem_final_exam"].shift(2)

# how many semesters the student has done before this one (gaps don't count)
sem_level["semesters_done"] = g["n_courses"].transform(lambda x: x.notna().cumsum().shift(1).fillna(0))

display(sem_level[sem_level["student_id"] == gap_student])

# looks right: the first semester has no previous value (NaN), and the semester after a gap gets
# NaN for the lag because the student really had nothing that semester. the rolling mean still
# uses whatever older semesters exist since we set min_periods=1.
#
# now we attach these back to the course rows. the gap rows just don't match any course so they
# don't get attached to anything.

lag_cols = ["final_exam_prev_semester", "homework_prev_semester", "final_exam_roll2",
            "final_exam_roll3", "final_exam_trend", "semesters_done", "n_courses", "credit_load"]

df = df.merge(sem_level[["student_id", "sem_idx"] + lag_cols], on=["student_id", "sem_idx"], how="left")
print(df.shape)
display(df[lag_cols].isna().mean().round(3))

# the lag columns have some NaNs, which is expected because nobody has a "previous semester" in
# their first one. we don't fill these here, the imputer inside the pipeline handles them (fit on
# train only). we also add a flag so the model knows if a student had a previous semester at all,
# since "no history" is info in itself.

df["has_prev_semester"] = df["final_exam_prev_semester"].notna().astype(int)


# ======================================================================
# 2.3 EXTRA DERIVED FEATURES
# ======================================================================
#
# a few features that make sense from an education point of view:
# - engagement = study_hours × attendance_rate / 100. someone who studies a lot and shows up is
#   different from someone who only does one of them
# - hw_vs_gpa = homework_avg minus prior_gpa scaled to 100. positive means they're doing better
#   this course than their usual level
# - heavy_load = 1 if the student's credit load that semester is higher than most (above the 75th
#   percentile, so 20+ credits)

df["engagement"] = df["study_hours"] * df["attendance_rate"] / 100
df["hw_vs_gpa"] = df["homework_avg"] - df["prior_gpa"] * 25
df["heavy_load"] = (df["credit_load"] > df["credit_load"].quantile(0.75)).astype(int)

display(df[["engagement", "hw_vs_gpa", "heavy_load", "credit_load"]].describe().T)


# ======================================================================
# 2.4 ENCODING THE CATEGORICAL COLUMNS
# ======================================================================
#
# choices we made and why:
# - year_of_study has a natural order (freshman < sophomore < junior < senior) so we use ordinal
#   encoding, one-hot would lose that order
# - subject_area, gender, instructor don't have an order so they get one-hot. instructor only has
#   20 values so that's fine
# - course_name is dropped, it's 1 to 1 with course_code so it's literally the same info twice
# - course_code has 40 values. we keep it one-hot since some courses are probably harder than
#   others, and 40 extra columns is still manageable
# - semester text is dropped since sem_idx already has it as a number
# - student_id is dropped, it's just an id. if we kept it the model could memorize students
#   instead of learning patterns
# - binary columns (scholarship, internship) are already 0/1 so nothing to do
# - numeric scaling is done with StandardScaler, but it's fit on the training set only inside the
#   pipeline so no info from the test set leaks in

ordinal_cols = ["year_of_study"]
year_order = [["Freshman", "Sophomore", "Junior", "Senior"]]
onehot_cols = ["subject_area", "gender", "instructor", "course_code"]

numeric_cols = ["credits", "prior_gpa", "study_hours", "attendance_rate", "age", "extracurricular",
                "homework_avg", "sem_idx", "final_exam_prev_semester", "homework_prev_semester",
                "final_exam_roll2", "final_exam_roll3", "final_exam_trend", "semesters_done",
                "n_courses", "credit_load", "engagement", "hw_vs_gpa"]
binary_cols = ["scholarship", "internship", "has_prev_semester", "heavy_load"]

def build_preprocessor(num_cols):
    numeric_pipe = Pipeline([
        ("impute", SimpleImputer(strategy="median")),   # mostly for the lag NaNs
        ("scale", StandardScaler())
    ])
    return ColumnTransformer([
        ("num", numeric_pipe, num_cols),
        ("ord", OrdinalEncoder(categories=year_order), ordinal_cols),
        ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), onehot_cols),
        ("bin", "passthrough", binary_cols)
    ])


# ======================================================================
# 2.5 TWO FEATURE SETS (REGRESSION AND CLASSIFICATION)
# ======================================================================
#
# regression (target = final_exam)
# - we drop the rows where final_exam is missing (we didn't fill the target, remember)
# - we leave out numeric_grade and grade_letter. the course grade is calculated from the final
#   exam (they correlate 0.66), so using it to predict the final would be cheating
#
# classification (target = grade_letter)
# - numeric_grade is out, it's basically the letter grade as a number (we showed the ranges in
#   1.3)
# - we keep final_exam as a feature since the brief only says to exclude numeric_grade. but it's
#   missing in about 7% of rows so it goes through the imputer too

reg_df = df.dropna(subset=["final_exam"]).copy()
X_reg = reg_df[numeric_cols + ordinal_cols + onehot_cols + binary_cols]
y_reg = reg_df["final_exam"]

clf_numeric_cols = numeric_cols + ["final_exam"]
X_clf = df[clf_numeric_cols + ordinal_cols + onehot_cols + binary_cols]
y_clf = df["grade_letter"]

print("regression X:", X_reg.shape, "| y:", y_reg.shape)
print("classification X:", X_clf.shape, "| y:", y_clf.shape)

# double check no leakage columns slipped in
for leak in ["numeric_grade", "grade_letter", "student_id"]:
    assert leak not in X_reg.columns and leak not in X_clf.columns
print("no leakage columns in either set")


# ======================================================================
# 2.6 CLASS IMBALANCE
# ======================================================================
#
# checking how balanced the letter grades are.

grade_order = ["A", "A-", "B+", "B", "B-", "C+", "C", "C-", "D", "F"]
counts = y_clf.value_counts().reindex(grade_order)

plt.figure(figsize=(8, 3.5))
sns.barplot(x=counts.index, y=counts.values, color="steelblue")
plt.title("grade_letter distribution (before balancing)")
plt.ylabel("count")
plt.show()

print(counts)
print("\nratio biggest / smallest class:", round(counts.max() / counts.min(), 1))

# yeah it's really imbalanced. B has around 27k rows but F has only ~540, so about 50 times less.
# if we leave it like this the model can basically ignore F and still get good accuracy, but F
# students are literally the ones we care about most (at-risk students).
#
# we fix it with SMOTE, which makes new synthetic samples for the small classes by mixing between
# real neighbours, instead of just copying the same rows over and over (that tends to overfit).
# the big rule here: SMOTE is applied only to the training set. the test set stays real and
# imbalanced, because that's what the model will actually see in real life.


# ======================================================================
# 2.7 TRAIN / TEST SPLIT
# ======================================================================
#
# 80/20 split. for classification we stratify on grade_letter so the test set has the same grade
# mix as the full data (otherwise the test set might end up with barely any F's). the
# preprocessor is fit on train only and then just applied to test.

# regression split
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=RANDOM_STATE)

reg_prep = build_preprocessor(numeric_cols)
Xr_train_ready = reg_prep.fit_transform(Xr_train)   # fit on train only
Xr_test_ready = reg_prep.transform(Xr_test)          # just apply to test

print("regression train:", Xr_train_ready.shape, "| test:", Xr_test_ready.shape)

# classification split (stratified)
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=RANDOM_STATE, stratify=y_clf)

clf_prep = build_preprocessor(clf_numeric_cols)
Xc_train_ready = clf_prep.fit_transform(Xc_train)
Xc_test_ready = clf_prep.transform(Xc_test)

print("classification train:", Xc_train_ready.shape, "| test:", Xc_test_ready.shape)

# SMOTE on the training data only
smote = SMOTE(random_state=RANDOM_STATE)
Xc_train_bal, yc_train_bal = smote.fit_resample(Xc_train_ready, yc_train)

before_after = pd.DataFrame({
    "train_before": yc_train.value_counts().reindex(grade_order),
    "train_after_smote": pd.Series(yc_train_bal).value_counts().reindex(grade_order),
    "test_untouched": yc_test.value_counts().reindex(grade_order)
})
display(before_after)

fig, axes = plt.subplots(1, 2, figsize=(12, 3.5), sharey=True)
before_after["train_before"].plot(kind="bar", ax=axes[0], color="indianred", title="train before SMOTE")
before_after["train_after_smote"].plot(kind="bar", ax=axes[1], color="seagreen", title="train after SMOTE")
plt.tight_layout()
plt.show()

# every grade has the same number of training rows now. the test set is untouched so the
# evaluation later is still honest.
#
# one thing to keep in mind for the report: SMOTE makes the training set way bigger (from ~121k
# to ~219k rows), so training will be slower. if that becomes a problem with the bigger models
# later we can try class_weight="balanced" instead, which does a similar job without making new
# rows.


# ======================================================================
# 2.8 SAVING THE PROCESSED DATA
# ======================================================================
#
# saving the cleaned data and the student-semester table so the next milestones (and the other
# group members) can just load them instead of rerunning everything.

df.to_csv("education_clean_features.csv", index=False)
sem_level.to_csv("student_semester_level.csv", index=False)
print("saved:", df.shape, sem_level.shape)

# milestone 2 summary:
# - semesters are in proper time order (sem_idx), each student is sorted by it, and missing
#   semesters in the middle are filled with NaN so the lags are always really "the semester
#   before"
# - we added lag features (final_exam_prev_semester, homework_prev_semester), rolling means over
#   2 and 3 semesters and a trend feature, all shifted so there's no leakage
# - we added some derived features (engagement, hw_vs_gpa, credit load, heavy_load)
# - categoricals are encoded (ordinal for year_of_study, one-hot for the rest), numbers are
#   scaled, and all of it is fit on train only
# - we have two clean feature sets: one for regression (final_exam) and one for classification
#   (grade_letter), with numeric_grade kept out of both
# - the class imbalance (B is ~50x bigger than F) is handled with SMOTE on the training set only
#
# next up is milestone 3: the correlation and time series analysis.
```
